```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
categorical_features = [
    "industry",
    "country",
    "referral_source",
    "plan_tier",
    "signup_day"
]

numerical_features = [
    "seats",
    "signup_year",
    "signup_month",
    "tenure_days"
]

print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
encoder.fit(X_train[categorical_features])
```

```python
X_train_encoded = encoder.transform(X_train[categorical_features])
X_test_encoded = encoder.transform(X_test[categorical_features])

print("X_train_encoded:", X_train_encoded.shape)
print("X_test_encoded:", X_test_encoded.shape)
```

```python
X_train_numeric = X_train[numerical_features].to_numpy()
X_test_numeric = X_test[numerical_features].to_numpy()

print("X_train_numeric:", X_train_numeric.shape)
print("X_test_numeric:", X_test_numeric.shape)
```

```python
import numpy as np

X_train_processed = np.hstack([
    X_train_numeric,
    X_train_encoded
])

X_test_processed = np.hstack([
    X_test_numeric,
    X_test_encoded
])

print("X_train_processed:", X_train_processed.shape)
print("X_test_processed:", X_test_processed.shape)
```

```python
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nChurn distribution in training data:")
print(y_train.value_counts())
```

```python
y_train_array = y_train.to_numpy()
y_test_array = y_test.to_numpy()

print("y_train_array:", y_train_array.shape)
print("y_test_array:", y_test_array.shape)
```

```python
print("Training data:", X_train_processed.shape, y_train_array.shape)
print("Testing data:", X_test_processed.shape, y_test_array.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
model.fit(
    X_train_processed,
    y_train_array
)
```

```python
y_pred = model.predict(X_test_processed)

print("Predictions:", y_pred[:10])
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test_array, y_pred)

print("Baseline Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test_array, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test_array, y_pred)
recall = recall_score(y_test_array, y_pred)

print("Precision:", precision)
print("Recall:", recall)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
rf_model.fit(
    X_train_processed,
    y_train_array
)
```

```python
rf_pred = rf_model.predict(X_test_processed)

print("Predictions:", rf_pred[:10])
print("Number of predictions:", len(rf_pred))
```

```python
rf_accuracy = accuracy_score(
    y_test_array,
    rf_pred
)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
rf_cm = confusion_matrix(
    y_test_array,
    rf_pred
)

print("Random Forest Confusion Matrix:")
print(rf_cm)
```

```python
rf_precision = precision_score(
    y_test_array,
    rf_pred
)

rf_recall = recall_score(
    y_test_array,
    rf_pred
)

print("Random Forest Precision:", rf_precision)
print("Random Forest Recall:", rf_recall)
```

```python
model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [accuracy, rf_accuracy],
    "Precision": [precision, rf_precision],
    "Recall": [recall, rf_recall]
})

model_comparison
```

```python
rf_probabilities = rf_model.predict_proba(X_test_processed)

print("Probability shape:", rf_probabilities.shape)
print(rf_probabilities[:5])
```

```python
churn_probability = rf_probabilities[:, 1]

print("First 10 churn probabilities:")
print(churn_probability[:10])
```

```python
risk_df = df.loc[
    X_test.index,
    [
        "account_id",
        "account_name",
        "industry",
        "country",
        "plan_tier",
        "seats",
        "is_trial"
    ]
].copy()

risk_df["churn_probability"] = churn_probability

risk_df = risk_df.sort_values(
    "churn_probability",
    ascending=False
)

risk_df.head(10)
```

```python
high_risk_customers = risk_df[
    risk_df["churn_probability"] >= 0.50
].copy()

print("Number of high-risk customers:", len(high_risk_customers))

high_risk_customers
```
